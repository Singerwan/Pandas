```python
import numpy as np
import pandas as pd
```

```python
df=pd.DataFrame(data    = np.random.randn(8,5),
                index   = map(lambda x:"Row_{}".format(x) , np.arange(1,9)),
                columns = map(lambda y:"Col_{}".format(y) , np.arange(1,6)) )
def indexstyle(df):
    return "border:dashed cyan 4px;font-size:20px;color:lightblue"
def columnstyle(df):
    return "border:dashed magenta 4px;font-size:20px;color:teal"

def max_val(x):
    if x in list(df.max()):
        return "color:red;font-size:25px;font-weight:bold;background-color:indigo;border:double white 10px;padding:20px"

def row_ind(x):
    if x in list(map(lambda x :"Row_{}".format(x) , np.arange(2,5))):
        return "background-color:cyan"

df.style.set_table_styles( [ {"selector":"",
                              "props":"border:solid brown 8px;padding:30px"} ] ).\
         set_properties(  **{"border":"dotted blue 4px"} ).\
         background_gradient(cmap="rainbow").\
         map_index(indexstyle,axis=1).\
         map_index(columnstyle,axis=0).map(max_val).map_index(row_ind)
```
